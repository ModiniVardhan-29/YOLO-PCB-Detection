```python
import os, shutil, random

BASE = "/kaggle/working/pcb_dataset"
NEW = "/kaggle/working/pcb_dataset_fixed"

splits = {"train": 0.8, "val": 0.1, "test": 0.1}

for s in splits:
    os.makedirs(f"{NEW}/{s}/images", exist_ok=True)
    os.makedirs(f"{NEW}/{s}/labels", exist_ok=True)

images = os.listdir(f"{BASE}/train/images")
random.shuffle(images)

n = len(images)
train_n = int(n * 0.8)
val_n = int(n * 0.1)

train_imgs = images[:train_n]
val_imgs = images[train_n:train_n + val_n]
test_imgs = images[train_n + val_n:]

def copy_files(img_list, split):
    for img in img_list:
        shutil.copy(f"{BASE}/train/images/{img}", f"{NEW}/{split}/images/{img}")
        lbl = img.replace(".jpg", ".txt").replace(".png", ".txt")
        shutil.copy(f"{BASE}/train/labels/{lbl}", f"{NEW}/{split}/labels/{lbl}")

copy_files(train_imgs, "train")
copy_files(val_imgs, "val")
copy_files(test_imgs, "test")

print("Dataset re-split complete!")
```

```python
import os, random, shutil

BASE = "/kaggle/working/pcb_dataset"

train_img = f"{BASE}/train/images"
train_lbl = f"{BASE}/train/labels"

val_img = f"{BASE}/val/images"
val_lbl = f"{BASE}/val/labels"

test_img = f"{BASE}/test/images"
test_lbl = f"{BASE}/test/labels"

# make sure folders exist
os.makedirs(val_img, exist_ok=True)
os.makedirs(val_lbl, exist_ok=True)
os.makedirs(test_img, exist_ok=True)
os.makedirs(test_lbl, exist_ok=True)

imgs = os.listdir(train_img)
random.seed(42)
random.shuffle(imgs)

n = len(imgs)
val_n = int(0.1 * n)
test_n = int(0.1 * n)

val_imgs = imgs[:val_n]
test_imgs = imgs[val_n:val_n + test_n]

def move(img_list, dst_img, dst_lbl):
    for img in img_list:
        lbl = img.replace(".jpg", ".txt").replace(".png", ".txt")
        shutil.move(f"{train_img}/{img}", f"{dst_img}/{img}")
        shutil.move(f"{train_lbl}/{lbl}", f"{dst_lbl}/{lbl}")

move(val_imgs, val_img, val_lbl)
move(test_imgs, test_img, test_lbl)

print("✔ Dataset split fixed successfully")
```

**supervised transfer learning**

```python
from ultralytics import YOLO
import torch, random, numpy as np

# -------------------------------
# Reproducibility (important)
# -------------------------------
seed = 21
torch.manual_seed(seed)
np.random.seed(seed)
random.seed(seed)

# -------------------------------
# Load pretrained YOLOv8n
# -------------------------------
model = YOLO("yolov8n.pt")

# -------------------------------
# Supervised Transfer Learning
# -------------------------------
results = model.train(
    data="/kaggle/working/data.yaml",

    epochs=100,            # 🔑 fewer epochs → faster
    imgsz=640,             # 🔑 balanced speed & accuracy
    batch=32,              # 🔑 faster on T4 GPU

    optimizer="AdamW",
    lr0=0.001,
    weight_decay=0.0005,
    cos_lr=True,

    patience=0,
    deterministic=True,

    # Moderate augmentation (generalization without slowdown)
    mosaic=0.3,
    mixup=0.1,
    fliplr=0.5,
    hsv_h=0.015,
    hsv_s=0.6,
    hsv_v=0.4,

    pretrained=True,
    name="pcb_transfer_learning_fast",
    verbose=True
)

print("Training complete.")
print("Run folder:", results.save_dir)
```

```python
from ultralytics import YOLO
import torch, random, numpy as np

# -------------------------------
# Reproducibility
# -------------------------------
seed = 33
torch.manual_seed(seed)
np.random.seed(seed)
random.seed(seed)

# -------------------------------
# Load YOLOv8n (same architecture)
# -------------------------------
model = YOLO("yolov8n.pt")

# -------------------------------
# HYBRID STRATEGY:
# Freeze backbone, train head
# -------------------------------
# Freeze early backbone layers (feature reuse)
for layer in model.model.model[:10]:
    for param in layer.parameters():
        param.requires_grad = False

# -------------------------------
# Hybrid Training
# -------------------------------
results = model.train(
    data="/kaggle/working/data.yaml",

    epochs=140,            # 🔑 deeper convergence
    imgsz=640,
    batch=32,

    optimizer="AdamW",
    lr0=0.001,
    weight_decay=0.0005,
    cos_lr=True,

    patience=20,
    deterministic=True,

    # Moderate augmentation (not heavy)
    mosaic=0.2,
    mixup=0.1,
    fliplr=0.5,
    hsv_h=0.015,
    hsv_s=0.6,
    hsv_v=0.4,

    pretrained=True,
    name="pcb_hybrid_final",
    verbose=True
)

print("Hybrid training complete.")
print("Saved at:", results.save_dir)
```
